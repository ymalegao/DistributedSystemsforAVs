```python
from typing import List
```

```python
R = [0, 3, 4, 0 , 5 , 2 , 7]
def minimize_broadcast(R: List , k: int):

   
    n = len(R) - 1
    print("Input List: ", R)
    requests = [0] * (n +1)
    sum_of_requests = [0] * ( n +1)
    dp = [[float('inf')] * (k + 1) for _ in range(n + 2)]
    parent = [[-1] * (k + 1) for _ in range(n + 2)]
    dp[0][0] = 0
    for i in range(1, n+1):
        requests[i] = requests[i-1] + R[i]
        sum_of_requests[i] = sum_of_requests[i-1] + i * R[i]
    print("Requests: ", requests)
    print("Sum of Requests: ", sum_of_requests)

    for i in range(1, k+1):
        for j in range(1,  n+1):

            best = float('inf')
            best_a = -1
            for a in range(i - 1, j):
                if a == 0:
                    continue
                cost = j * (requests[j-1] - requests[a-1]) - (sum_of_requests[j-1] - sum_of_requests[a-1])
                if dp[a][i-1] + cost < best:
                    best = dp[a][i-1] + cost
                    best_a = a
            dp[j][i] = best
            parent[j][i] = best_a
            print(f"DP[{j}][{i}] updated to: ", dp[j][i], " with partition at: ", best_a)

    print("DP Table: ")
    best_j = 0
    best_val = float('inf')
    for j in range(1, n+2):
        if dp[j][k] < best_val:
            best_val = dp[j][k]
            best_j = j

    sched = [ ]
    i, j = best_j, k
    while i > 0 and j > 0:
        sched.append(parent[i][j])
        i = parent[i][j]
        j -= 1
    sched.reverse()
    print("Optimal Broadcast Schedule (partitions at indices): ", sched)
    
    
    
```

```python
minimize_broadcast(R, 3)
```
